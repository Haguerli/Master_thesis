```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
h = 12.7
l = 21 # np.linspace(0, 50, 100) + 1 
maxangle = np.arctan(h/l)
minangle = np.arctan(h/(l+50))
print("maxangle: ", maxangle, "minangle: ", minangle)

angles = np.linspace(0, maxangle*1.1, 1000)
def effiency(alpha, l, h, maxangle, minangle):
    if alpha > maxangle:
        return 0
    elif alpha < minangle:
        return 0.85
    else:
        print(h-np.tan(alpha)*l)
        return (((h-np.tan(alpha)*l) / np.cos(np.pi/2-alpha))/60)*0.85



value = []
for a in angles:
    value.append(effiency(a, l, h, maxangle, minangle))
plt.plot(angles, value)
plt.show()
    

```
